## Conjuntos en Python

- Python tiene <u>nativamente integrado</u> el tipo de dato `set`, el cual nos permite representar conjuntos y realizar sus operaciones.
- A diferencia de las listas o tuplas, los conjuntos <u>no permiten elementos duplicados</u> y <u>no tienen un orden definido</u>.

```python
conjunto_a = {1, 2, 3, 4, 5}
conjunto_b = set(["A", "B", "C", "D", "E"])

print(type(conjunto_a))
print(type(conjunto_b))
```

```python
# @title Añadir elementos

conjunto_a.add(6)
conjunto_b.update(["F", "G"])

print(conjunto_a)
print(conjunto_b)
```

```python
# @title Eliminar elementos

conjunto_a.remove(3)
conjunto_b.discard("B")

print(conjunto_a)
print(conjunto_b)
```

```python
# @title Unión de conjuntos

A = {"fútbol", "música", "viajes", "fotografía", "cocina"}
B = {"música", "arte", "lectura", "baile", "teatro"}

union = A.union(B)
print(f'A unión B: {union}')
```

```python
# @title Intersección de conjuntos

intereseccion = A.intersection(B)
print(f'A intersección B: {intereseccion}')
```

```python
# @title Diferencia de conjuntos

a_diferencia_b = A.difference(B)
print("A - B:", a_diferencia_b)
```

```python
# @title Diferencia simétrica de conjuntos

diferencia_simetrica = A.symmetric_difference(B)
print(f'A diff simetrica B: {diferencia_simetrica}')
```

```python
# @title Diferencia simétrica de conjuntos

# agregamos otro conjunto
C = {"cine", "arte", "música", "deportes", "cocina"}

# definimos el conjunto universal
conjunto_universal = A.union(B).union(C)

# calculamos el complemento de A con respecto al conjunto universal
complemento_a = conjunto_universal - A

print("Complemento de A:", complemento_a)
```

```python
# @title Llamado alternativo a operaciones

A = {1, 2, 3, 4, 5}
B = {6, 7, 8, 9, 10}

# unión de conjuntos
union = A | B

# intersección de conjuntos
interseccion = A & B

# diferencia entre conjuntos
diferencia = A - B

# diferencia simétrica entre conjuntos
diferencia_simetrica = A ^ B
```
